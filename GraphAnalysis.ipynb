## Generate Graph

```python
import networkx as nx
import json

# Load your edges from JSON or define them here
with open("mention_networks/network_mega_merge.json", "r", encoding="utf-8") as f:
    edges = json.load(f)

# Build directed graph
G = nx.DiGraph()
G.add_edges_from(edges)

source_nodes = ["soyvixterdev", "wgpaulo", "tobyfox", "rubiu5", "wizkhalifa", 
                "eldonutjaja", "guillermodbg21", "systemofadown", "quantumfracture", 
                "kojima_hideo", "serjtankian"]
```

## Find Longest Shortest Path

```python
def find_longest_shortest_path(G, source_nodes):
    max_distance = -1
    max_path = []

    for source in source_nodes:
        lengths, paths = nx.single_source_shortest_path_length(G, source), nx.single_source_shortest_path(G, source)
        for target, dist in lengths.items():
            if dist > max_distance:
                max_distance = dist
                max_path = paths[target]

    return max_path, max_distance

max_path, max_distance = find_longest_shortest_path(G, source_nodes)

print(max_path)
print(max_distance)
```

## Find K Longest paths

```python
import heapq
import networkx as nx

def find_top_k_longest_shortest_paths(G, source_nodes, k=10):
    top_k = []  # Min-heap to keep top-k (distance, path) tuples

    for source in source_nodes:
        lengths = nx.single_source_shortest_path_length(G, source)
        paths = nx.single_source_shortest_path(G, source)

        for target, dist in lengths.items():
            if source == target:
                continue
            path = paths[target]
            entry = (dist, path)
            if len(top_k) < k:
                heapq.heappush(top_k, entry)
            else:
                if dist > top_k[0][0]:
                    heapq.heappushpop(top_k, entry)

    # Sort results from longest to shortest
    return sorted(top_k, key=lambda x: -x[0])

source_nodes2 = ["delta"]

top_paths = find_top_k_longest_shortest_paths(G, source_nodes2, 10)


for i, (length, path) in enumerate(top_paths, 1):
    print(f"{i}. Length: {length}, Path: {path}")
```
