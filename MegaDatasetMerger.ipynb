```python
import os
import json

def merge_edge_jsons(folder_path, output_filename="merged_edges.json"):
    edge_set = set()

    for filename in os.listdir(folder_path):
        if filename.endswith(".json"):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, "r", encoding="utf-8") as f:
                try:
                    edges = json.load(f)
                    for edge in edges:
                        if isinstance(edge, list) and len(edge) == 2:
                            # Normalize both usernames to lowercase
                            user1, user2 = edge[0].lower(), edge[1].lower()
                            if user1 != user2:  # Skip self-edges
                                edge_set.add((user1, user2))
                except json.JSONDecodeError:
                    print(f"⚠️ Skipped invalid JSON: {filename}")

    # Convert set back to list of lists
    merged_edges = [list(edge) for edge in edge_set]

    output_path = os.path.join(folder_path, output_filename)
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(merged_edges, f, indent=2)

    print(f"✅ Merged {len(edge_set)} directed edges (lowercased, no self-edges) into {output_path}")

# Example usage
merge_edge_jsons("mention_networks", "network_mega_merge.json")
```
